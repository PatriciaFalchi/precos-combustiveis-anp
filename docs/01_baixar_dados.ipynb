```python
from google.colab import drive
drive.mount('/content/drive')

import requests, pathlib

PASTA = pathlib.Path('/content/drive/MyDrive/precos-combustiveis-anp/dados_brutos')
PASTA.mkdir(parents=True, exist_ok=True)

BASE = ("https://www.gov.br/anp/pt-br/centrais-de-conteudo/"
        "dados-abertos/arquivos/shpc/dsas/ca/")
nomes = [f"ca-{ano}-{sem:02d}" for ano in (2023, 2024, 2025) for sem in (1, 2)]
nomes.append("ca-2026-01")

for nome in nomes:
    for ext in (".zip", ".csv"):          # alguns anos podem ser csv
        r = requests.get(BASE + nome + ext, timeout=120,
                         headers={"User-Agent": "Mozilla/5.0"})
        if r.status_code == 200:
            (PASTA / (nome + ext)).write_bytes(r.content)
            print("ok ", nome + ext, round(len(r.content)/1e6, 1), "MB")
            break
    else:
        print("FALHOU", nome)
```
